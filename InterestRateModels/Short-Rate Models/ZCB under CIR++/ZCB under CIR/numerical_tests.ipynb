```python
from zcb_pricer import*
from callable_zcb_pricer import*
from bond_sensitivities import*
import matplotlib.pyplot as plt
```

## Parameters

```python
# CIR parameters
r0 = 0.05
a = 0.1
b = 0.1
sigma = 0.1

# Bond parameters
N = 100
Kc = 0.95 * N
T = 1       # maturity in years
t_call_eur = 8/12 * T # European call date
tc = t_call_eur
t_call_ber = np.arange(1/12, T, 1/12)  # Bermudan monthly call dates

# Simulation parameters
n_time, n_sim = 500, 50_000

# Finite Difference parameters
Nt, Nr = 150, 200
```

## Numerical Results

### First Results

```python
zcb_cir_analytical(r0, a, b, sigma, T, N), zcb_cir_cn(r0, a, b, sigma, T, Nt, Nr, N)#, zcb_cir_mc(r0, a, b, sigma, T, n_time, n_sim, N)
```

```python
# Single r0 comparison
r0_val = r0  # 0.05

# Vanilla ZCB (analytical)
price_vanilla = zcb_cir_analytical(r0_val, a, b, sigma, T, N)

# European callable
price_eur_cn = callable_zcb_cir_cn(
    r0_val, a, b, sigma, T, t_call_eur,
    Nt=Nt, Nr=Nr, N=N, Kc=Kc, exercise_type="European"
)
price_eur_lsm = callable_zcb_cir_lsm(
    r0_val, a, b, sigma, T, t_call_eur,
    n_time=n_time, n_sim=n_sim, N=N, Kc=Kc, exercise_type="European"
)

# Bermudan callable
price_ber_cn = callable_zcb_cir_cn(
    r0_val, a, b, sigma, T, t_call_ber,
    Nt=Nt, Nr=Nr, N=N, Kc=Kc, exercise_type="Bermudan"
)
price_ber_lsm = callable_zcb_cir_lsm(
    r0_val, a, b, sigma, T, t_call_ber,
    n_time=n_time, n_sim=n_sim, N=N, Kc=Kc, exercise_type="Bermudan"
)

# Print nicely in two sides
print(f"{'Vanilla ZCB':<20} {price_vanilla:>10.4f}")
print("\nEuropean Callable:")
print(f"{'CN':<10} {price_eur_cn:>10.4f}")
print(f"{'LSM':<10} {price_eur_lsm:>10.4f}")

print("\nBermudan Callable:")
print(f"{'CN':<10} {price_ber_cn:>10.4f}")
print(f"{'LSM':<10} {price_ber_lsm:>10.4f}")
```

```python
# Non-callable ZCB
D_nc, C_nc = bond_risk_measures(zcb_cir_analytical, r0, a, b, sigma, T, N)

# European callable ZCB
D_eur, C_eur = bond_risk_measures(
    callable_zcb_cir_cn, r0, a, b, sigma, T, t_call_eur,
    Nt=Nt, Nr=Nr, N=N, Kc=Kc, exercise_type="European"
)

# Bermudan callable ZCB
D_ber, C_ber = bond_risk_measures(
    callable_zcb_cir_cn, r0, a, b, sigma, T, t_call_ber,
    Nt=Nt, Nr=Nr, N=N, Kc=Kc, exercise_type="Bermudan"
)
```
