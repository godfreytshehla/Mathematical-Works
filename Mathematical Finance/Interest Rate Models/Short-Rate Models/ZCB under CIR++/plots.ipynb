```python
import numpy as np
import matplotlib.pyplot as plt

# ====================== Parameters ======================
X = 10.0      # Bonus Level
H = 6.0       # Downside Barrier Level

S_T = np.linspace(0, 20, 500)

# ====================== Payoff ======================
payoff_bonus = np.maximum(S_T, X)           # Barrier not hit
payoff_hit   = S_T                          # Barrier hit

# ====================== Plot ======================
plt.figure(figsize=(11, 7))

plt.plot(S_T, payoff_bonus, 
         label=r'$S^{min}_T > H$ (Bonus Active)', 
         color='blue', linewidth=2.5)

plt.plot(S_T, payoff_hit, 
         label=r'$S^{min}_T \leq H$ (Barrier Hit)', 
         color='red', linestyle='--', linewidth=2.2)

# Reference lines
plt.axhline(y=X, color='gray', linestyle=':', alpha=0.7, 
            label=f'Bonus Level (X = {X})')
plt.axvline(x=H, color='purple', linestyle=':', alpha=0.7, 
            label=f'Barrier (H = {H})')
plt.axvline(x=X, color='green', linestyle=':', alpha=0.6)

plt.title('Standard Bonus Certificate Payoff at Maturity', 
          fontsize=14, fontweight='bold')
plt.xlabel('Terminal Stock Price $S_T$', fontsize=12)
plt.ylabel('Payoff $L_T$', fontsize=12)

plt.grid(True, alpha=0.3)
plt.legend(fontsize=11, loc='upper left')

# Annotations
plt.annotate('Bonus Region\n(max(S_T, X))', xy=(13, 14), xytext=(14, 17),
             arrowprops=dict(arrowstyle='->', color='blue'), fontsize=10, color='blue')

plt.annotate('Linear Payoff\n(Barrier Hit)', xy=(3, 3), xytext=(1, 8),
             arrowprops=dict(arrowstyle='->', color='red'), fontsize=10, color='red')

plt.tight_layout()
plt.show()
```
