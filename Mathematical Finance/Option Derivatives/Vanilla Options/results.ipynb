```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import importlib
import bsm_pricer
importlib.reload(bsm_pricer);
#importlib.reload(bsm_helpers);
```

```python
params = bsm_pricer.OptionParams(S0=10, X=10, T=9/12, r=0.05, q=0.03, sigma=0.2)

n_time = 252
n_sim = 10_000
n_t = 1_000
theta=0.95
N_t = 250
N_s = 200
```

```python
call_mc = bsm_pricer.monte_carlo_option_price(params, n_time=n_time, n_sim=n_sim, option_type="call")
put_mc = bsm_pricer.monte_carlo_option_price(params, n_time=n_time, n_sim=n_sim, option_type="put")

call_bsm = bsm_pricer.analytical_bsm_option_price(params, option_type="call")
put_bsm = bsm_pricer.analytical_bsm_option_price(params, option_type="put")

call_bin = bsm_pricer.binomial_option_price(params, n_t=n_t, option_type="call")
put_bin = bsm_pricer.binomial_option_price(params, n_t=n_t, option_type="put")

call_fdm = bsm_pricer.bsm_option_price_fdm(params, N_t=N_t, N_s=N_s, option_type="call")
put_fdm = bsm_pricer.bsm_option_price_fdm(params, N_t=N_t, N_s=N_s, option_type="put")

call_theta = bsm_pricer.bsm_option_price_fdm(params, N_t=N_t, N_s=N_s, option_type="call", fdm_type="theta_imex", theta=theta)
put_theta = bsm_pricer.bsm_option_price_fdm(params, N_t=N_t, N_s=N_s, option_type="put", fdm_type="theta_imex", theta=theta)

call_int = bsm_pricer.bsm_option_price_integration(params, option_type="call")
put_int = bsm_pricer.bsm_option_price_integration(params, option_type="put")
```

```python
results = pd.DataFrame({
    "Analytical": [call_bsm, put_bsm],
    "Monte Carlo": [call_mc, put_mc],
    "Binomial": [call_bin, put_bin],
    "FDM": [call_fdm, put_fdm],
    "IMEX": [call_theta, put_theta],
    "NIntegration": [call_int, put_int]
}, index=["Call", "Put"])

results = results.round(6)

results
```
